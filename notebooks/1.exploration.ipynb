# 2. Exploration des données

```python
import pandas as pd

# Chargement du dataset
df = pd.read_csv("../data/weatherAUS.csv")
df_numeric = df.select_dtypes(include=["number"])
print("Données chargées avec succès.")
print("Dimensions du dataset :", df.shape)
```

## 2.1 Exploration du dataset

```python
print("Première visualisation :\n")
print(df.head())
```

```python
print("Types de variables :\n")
print(df.info())
```

```python
print("Statistiques descriptives :\n")
print(df_numeric.describe())
```

```python
print(f"\nPrésence de doublons : {df.duplicated().sum()}")
```

```python
print("Valeurs manquantes :\n")
print(df.isnull().sum())
```

```python
print("Valeurs manquantes (%) :\n")
print((df.isna().mean() * 100).round(2))
```

## 2.2 Distribution des variables

```python
print("\nListe des différentes Locations :\n")
print(df['Location'].unique())
```

```python
print("\nDistribution de Location (%) :\n")
print(df["Location"].value_counts(normalize=True).round(4) * 100)
```

```python
print("\nDistribution de RainToday (%) :\n")
print(df["RainToday"].value_counts(normalize=True).round(4) * 100)

print("\nDistribution de RainTomorrow (%) :\n")
print(df["RainTomorrow"].value_counts(normalize=True).round(4) * 100)
```

```python
print("\nDistribution de WindGustDir :\n")
print(df["WindGustDir"].value_counts(normalize=True).round(4) * 100)
```
