### Dummy Classifier for Weather Prediction

This section will demonstrate how to create and evaluate a `DummyClassifier` as a baseline for comparison with more complex models.

The target variable for this dataset is typically 'RainTomorrow'. We will separate the features (X) from the target (y) and then split the data into training and testing sets.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
import pandas as pd
X_train = pd.read_csv('../data/weatherAUS_X_train.csv')
X_test = pd.read_csv('../data/weatherAUS_X_test.csv')
y_train = pd.read_csv('../data/weatherAUS_y_train.csv').squeeze()
y_test = pd.read_csv('../data/weatherAUS_y_test.csv').squeeze()

print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")
```

Now, we will train a `DummyClassifier`. We'll use the 'stratified' strategy, which makes predictions by respecting the training set's class distribution. This is a common baseline for imbalanced datasets.

```python
# Initialize the DummyClassifier with a 'stratified' strategy
dummy_clf = DummyClassifier(strategy="stratified", random_state=42)

# Train the dummy classifier
dummy_clf.fit(X_train, y_train)

# Make predictions on the test set
y_pred_dummy = dummy_clf.predict(X_test)

# Evaluate the dummy classifier
accuracy_dummy = accuracy_score(y_test, y_pred_dummy)
report_dummy = classification_report(y_test, y_pred_dummy)

print(f"Dummy Classifier Accuracy: {accuracy_dummy:.4f}")
print("\nDummy Classifier Classification Report:")
print(report_dummy)
```

This `DummyClassifier` provides a baseline accuracy and F1-score that any more sophisticated model should aim to surpass. If your advanced model performs worse than this dummy model, it suggests issues with the model or data preprocessing.

### Detailed Evaluation Metrics for Dummy Classifier

Let's extract and calculate the requested metrics:
- **F1-macro**: The harmonic mean of precision and recall for each class, averaged across classes.
- **Recall (True/Yes)**: The ability of the classifier to find all the positive samples.
- **Precision (True/Yes)**: The ability of the classifier not to label as positive a sample that is negative.
- **AUC-ROC (Area Under the Receiver Operating Characteristic Curve)**: Measures the ability of a classifier to distinguish between classes. Higher AUC means better performance.
- **PR-AUC (Area Under the Precision-Recall Curve)**: Particularly useful for imbalanced datasets, as it focuses on the performance of the positive class.

```python
from sklearn.metrics import roc_auc_score, average_precision_score

# 1. Générer les probabilités du classifieur
y_proba_dummy = dummy_clf.predict_proba(X_test)

# 2. Trouver l'index de la classe positive (True)
dummy_classes = dummy_clf.classes_
positive_class_index = list(dummy_classes).index(True)
y_proba_positive = y_proba_dummy[:, positive_class_index]

# 3. Calculer l'AUC-ROC et le PR-AUC directement avec y_test (converti en booléen au besoin)
# Si y_test est sous forme de chaîne de caractères ("True"/"False"), utilisez : y_test.astype(bool)
auc_roc_dummy = roc_auc_score(y_test, y_proba_positive)
pr_auc_dummy = average_precision_score(y_test, y_proba_positive)

print(f"\n--- Metrics Dummy Classifier ---")
print(f"AUC-ROC: {auc_roc_dummy:.4f}")
print(f"PR-AUC: {pr_auc_dummy:.4f}")
```
